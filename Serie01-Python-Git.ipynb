```python
#load the libraries 
import numpy as np
import matplotlib.pyplot as plt

#load the data
ecg_raw = np.loadtxt('data/ECGu.txt')
print(ecg_raw.shape)
plt.plot(ecg_raw)
```

```python
#plotthe data 
plt.plot(ecg_raw)
plt.title('ECG signal')
plt.xlabel('samples')
plt.ylabel('amplitude')
plt.grid()
plt.show()
```

```python
# create t the time vector in seconds``
sampling_frequency = 1000 #in HZ
T = 1/ sampling_frequency
N = len(ecg_raw) #number of samples 
start = 0 
stop = (N - 1) * T #sampling frequency
t = np.linspace(start, stop, N)

#plot the data with time vector 
plt.plot(t, ecg_raw)
plt.title('ECG signal')
plt.xlabel('time (s)')
plt.ylabel('amplitude') 
plt.grid()
plt.show()  

```

```python
#rescal the ECG signal to mV knowing that 1 unit = 1024 microV
ecg = ecg_raw * 1024 / 1e-6 #in mV

#plot the data with time vector
plt.plot(t, ecg)
plt.title('ECG signal')
plt.xlabel('time (s)')
plt.ylabel('amplitude (mV)')
plt.grid()
plt.show() 
```
